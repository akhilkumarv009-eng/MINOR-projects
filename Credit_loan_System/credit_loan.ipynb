```python
import pandas as pd
```

```python
dataset=pd.read_csv("loan.csv")
dataset
dataset.info()
```

## HANDLING MISSING VALUES

```python
from sklearn.impute import SimpleImputer
```

```python
categorical_columns=dataset.select_dtypes(include=["str"]).columns
numerical_columns=dataset.select_dtypes(include=["float64"]).columns
categorical_columns
```

```python
num_imp=SimpleImputer(strategy="mean")
dataset[numerical_columns]=num_imp.fit_transform(dataset[numerical_columns]);

cat_imp=SimpleImputer(strategy="most_frequent")
dataset[categorical_columns]=cat_imp.fit_transform(dataset[categorical_columns]);

dataset
```

```python
dataset=dataset.drop("Applicant_ID",axis=1)
```

# feature encoding

```python
dataset.info()
```

```python
from sklearn.preprocessing import OneHotEncoder,LabelEncoder
```

```python
categorical_columns=dataset.select_dtypes(include=["str"]).columns
categorical_columns
```

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
import pandas as pd

# Columns for One-Hot Encoding
columns_OneHot = [
    "Employment_Status",
    "Property_Area",
    "Gender",
    "Employer_Category",
    "Loan_Approved",
    "Marital_Status"
]

# Label Encoder
LE = LabelEncoder()

dataset["Education_Level"] = LE.fit_transform(dataset["Education_Level"])
dataset["Loan_Purpose"] = LE.fit_transform(dataset["Loan_Purpose"])

# One-Hot Encoder
OE = OneHotEncoder(
    drop="first",
    sparse_output=False,
    handle_unknown="ignore"
)

# Transform the selected columns
encoded_array = OE.fit_transform(dataset[columns_OneHot])

# Create DataFrame of encoded columns
encoded_dataset = pd.DataFrame(
    encoded_array,
    columns=OE.get_feature_names_out(columns_OneHot),
    index=dataset.index
)

# Drop original columns and concatenate encoded columns
dataset = pd.concat(
    [dataset.drop(columns=columns_OneHot), encoded_dataset],
    axis=1
)

# Display first few rows
dataset.head()
```

```python
dataset.info()
```

## Corelation Matric

```python
nums_columns=dataset.select_dtypes(include="number");
correleation_mat=nums_columns.corr()["Loan_Approved_Yes"].sort_values(ascending=False)
correleation_mat
```

## test train and split

```python
X=dataset.drop(columns=["Loan_Approved_Yes"])
Y=dataset["Loan_Approved_Yes"]
X
```

```python
Y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(
    X,Y,test_size=0.2,random_state=42
)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

## training and evaluating about model

```python
 #logistic regression 
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,precision_score,recall_score,confusion_matrix
model1=LogisticRegression(max_iter=5000)

model1.fit(X_train_scaled,Y_train)

Y_predict=model1.predict(X_test_scaled)

print("accuracy-",accuracy_score(Y_test,Y_predict))
print("recall-",recall_score(Y_test,Y_predict))
print("precison-",precision_score(Y_test,Y_predict))
print(confusion_matrix(Y_test,Y_predict))
```

```python
#KNN
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score,precision_score,recall_score,confusion_matrix
model2=KNeighborsClassifier(n_neighbors=5)

model2.fit(X_train_scaled,Y_train)

Y_predict=model2.predict(X_test_scaled)

print("accuracy-",accuracy_score(Y_test,Y_predict))
print("recall-",recall_score(Y_test,Y_predict))
print("precison-",precision_score(Y_test,Y_predict))
print(confusion_matrix(Y_test,Y_predict))
```

```python
#naive_bayes
#KNN
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score,precision_score,recall_score,confusion_matrix
model3=GaussianNB()
model3.fit(X_train_scaled,Y_train)

Y_predict=model3.predict(X_test_scaled)

print("accuracy-",accuracy_score(Y_test,Y_predict))
print("recall-",recall_score(Y_test,Y_predict))
print("precison-",precision_score(Y_test,Y_predict))
print(confusion_matrix(Y_test,Y_predict))
```
